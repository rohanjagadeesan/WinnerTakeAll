```python
high_bins = [ [60,71], [71,78], [78,91], [91,97], [97,1000] ]
inclusive_bins = [ [0,25], [25,36], [36,48], [48,60]  ]
analysis_bins = [ [0,25], [25,36], [36,48], [48,60], [60,71], [71,78], [78,91], [91,97], [97,1000] ]
```

```python
ROOT.gDirectory.Clear()
tfile = ROOT.TFile.Open(filepath, "READ")

wta_profiles = {}
std_profiles = {}
EECs = {} # using standard

for mult_bin in high_bins:
    bin_key = f"{mult_bin[0]} < Nch < {mult_bin[1]}"
    bin_name = f"{mult_bin[0]}_{mult_bin[1]}"

    wta_profiles[bin_key] = ROOT.TGraphErrors( tfile.Get(f"WTA_profile_{bin_name}").Clone() )
    std_profiles[bin_key] = ROOT.TGraphErrors( tfile.Get(f"STD_profile_{bin_name}").Clone() )
    EECs[bin_key] = ROOT.TGraphErrors( tfile.Get(f"STD_eec_{bin_name}").Clone() )

    # detach them
    #wta_profiles[bin_key].SetDirectory(0)
    #std_profiles[bin_key].SetDirectory(0)
    #EECs[bin_key].SetDirectory(0)
```

```python
colours = [ROOT.kBlue+1, ROOT.kMagenta+1, ROOT.kSpring+2, ROOT.kRed+1, ROOT.kOrange+7]
```

```python


# pick out the TGraphErrors
eec_60_71 = EECs["60 < Nch < 71"]
eec_71_78 = EECs["71 < Nch < 78"]
eec_78_91 = EECs["78 < Nch < 91"]
eec_91_97 = EECs["91 < Nch < 97"]
eec_97_1000 = EECs["97 < Nch < 1000"]

# Choosing colours:
eec_60_71.SetLineColorAlpha(colours[0], 0.7)
eec_71_78.SetLineColorAlpha(colours[1], 0.7)
eec_78_91.SetLineColorAlpha(colours[2], 0.7)
eec_91_97.SetLineColorAlpha(colours[3], 0.7)
eec_97_1000.SetLineColorAlpha(colours[4], 0.7)

eec_60_71.SetMarkerColorAlpha(colours[0], 0.7)
eec_71_78.SetMarkerColorAlpha(colours[1], 0.7)
eec_78_91.SetMarkerColorAlpha(colours[2], 0.7)
eec_91_97.SetMarkerColorAlpha(colours[3], 0.7)
eec_97_1000.SetMarkerColorAlpha(colours[4], 0.7)

# if you want markers:
#eec_60_71.SetMarkerSize(0.4)
#eec_71_78.SetMarkerSize(0.4)
#eec_78_91.SetMarkerSize(0.4)
#eec_91_97.SetMarkerSize(0.4)
#eec_97_1000.SetMarkerSize(0.4)

# if you don't want markers:
eec_60_71.SetMarkerStyle(1)
eec_71_78.SetMarkerStyle(1)
eec_78_91.SetMarkerStyle(1)
eec_91_97.SetMarkerStyle(1)
eec_97_1000.SetMarkerStyle(1)
eec_60_71.SetMarkerSize(0)
eec_71_78.SetMarkerSize(0)
eec_78_91.SetMarkerSize(0)
eec_91_97.SetMarkerSize(0)
eec_97_1000.SetMarkerSize(0)

# initialise canvas
canvas = ROOT.TCanvas("c_eec", "EECs for different multiplicity bins", 750, 650)

canvas.SetLeftMargin(0.15)
canvas.SetBottomMargin(0.15)
canvas.SetGrid()

# initialise multigraph
mg = ROOT.TMultiGraph()

# Add the data markers
#mg.Add(eec_60_71, "L")
#mg.Add(eec_71_78, "L")
#mg.Add(eec_78_91, "L")
#mg.Add(eec_91_97, "L")
#mg.Add(eec_97_1000, "L")

# no error bars version:
mg.Add(eec_60_71, "LX")
mg.Add(eec_71_78, "LX")
mg.Add(eec_78_91, "LX")
mg.Add(eec_91_97, "LX")
mg.Add(eec_97_1000, "LX")

# title
mg.SetTitle("EEC for high multiplicity bins; #Delta R_{L}; E2C")
mg.Draw("AL") 

# Log scale axes
canvas.SetLogx(1)
canvas.SetLogy(1)
mg.GetXaxis().SetLimits(0.001, 1.0)
mg.GetYaxis().SetLimits(0.0001, 1.0)
#mg.GetXaxis().SetLimits(0, 1.0)
#mg.GetYaxis().SetLimits(0, 1.0)

mg.GetXaxis().SetTitleSize(0.045)
mg.GetYaxis().SetTitleSize(0.045)
mg.GetXaxis().SetLabelSize(0.04)
mg.GetYaxis().SetLabelSize(0.04)

# Legend setup
legend = ROOT.TLegend(0.48, 0.68, 0.88, 0.88)
legend.SetBorderSize(0)
legend.SetFillStyle(0) 
legend.SetTextSize(0.035)
legend.AddEntry(eec_60_71, "60 #leq N_{ch} < 71", "L")
legend.AddEntry(eec_71_78, "71 #leq N_{ch} < 78", "L")
legend.AddEntry(eec_78_91, "78 #leq N_{ch} < 91", "L")
legend.AddEntry(eec_91_97, "91 #leq N_{ch} < 97", "L")
legend.AddEntry(eec_97_1000, "97 #leq N_{ch} < 1000", "L")
legend.Draw()

canvas.Update()
canvas.SaveAs("EEC_plot.pdf")
canvas.Close()
```
